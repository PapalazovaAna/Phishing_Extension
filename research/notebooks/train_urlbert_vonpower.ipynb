```python

import random
import time
import dill
import numpy as np
import torch

from torch.utils.data import (
    DataLoader,
    IterableDataset,
)
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer,
)


MODEL_NAME = (
    "CrabInHoney/"
    "urlbert-tiny-v4-phishing-classifier"
)

MAX_LENGTH = 64
CHUNK_OVERLAP = 16
BATCH_SIZE = 32
EPOCHS = 1
LEARNING_RATE = 2e-5
NUM_WORKERS = 2
SEED = 42


random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)

print("Loading .dill dataset...")

with DRIVE_DATASET.open("rb") as file:
    data = dill.load(file)

train_x = np.asarray(data["train_x"])
train_y = np.asarray(data["train_y"]).reshape(-1)
char_to_int = data["char_to_int"]

reverse_char_to_int = {
    int(value): str(key)
    for key, value in char_to_int.items()
}

print("Train records:", len(train_x))
print("Train labels:", np.bincount(train_y))
print("Train shape:", train_x.shape)

print("Loading URLBERT tokenizer and model...")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

model = (
    AutoModelForSequenceClassification
    .from_pretrained(MODEL_NAME)
)

model.to(device)
model.train()

print("Model loaded.")
```

```python
def decode_url(row):
    characters = []

    for token in row:
        token = int(token)

        if token == 0:
            continue

        characters.append(
            reverse_char_to_int.get(token, "")
        )

    return "".join(characters)


def create_chunks(url):
    tokens = tokenizer.tokenize(url)

    token_ids = tokenizer.convert_tokens_to_ids(
        tokens
    )

    content_length = MAX_LENGTH - 2

    if len(token_ids) <= content_length:
        chunks = [token_ids]
    else:
        step = content_length - CHUNK_OVERLAP
        chunks = []

        for start in range(
            0,
            len(token_ids),
            step,
        ):
            chunk = token_ids[
                start:start + content_length
            ]

            if chunk:
                chunks.append(chunk)

            if start + content_length >= len(
                token_ids
            ):
                break

    result = []

    for chunk in chunks:
        input_ids = [
            tokenizer.cls_token_id,
            *chunk,
            tokenizer.sep_token_id,
        ]

        attention_mask = [1] * len(input_ids)

        padding_length = (
            MAX_LENGTH - len(input_ids)
        )

        input_ids.extend(
            [tokenizer.pad_token_id or 0]
            * padding_length
        )

        attention_mask.extend(
            [0] * padding_length
        )

        result.append(
            (
                input_ids,
                attention_mask,
            )
        )

    return result


class VonpowerChunkDataset(IterableDataset):
    def __init__(self, indices):
        self.indices = indices

    def __iter__(self):
        worker_info = (
            torch.utils.data.get_worker_info()
        )

        indices = self.indices

        if worker_info is not None:
            indices = indices[
                worker_info.id::worker_info.num_workers
            ]

        for index in indices:
            url = decode_url(train_x[index])
            label = int(train_y[index])

            chunks = create_chunks(url)

            for input_ids, attention_mask in chunks:
                yield {
                    "input_ids": torch.tensor(
                        input_ids,
                        dtype=torch.long,
                    ),
                    "attention_mask": torch.tensor(
                        attention_mask,
                        dtype=torch.long,
                    ),
                    "labels": torch.tensor(
                        label,
                        dtype=torch.long,
                    ),
                }
```

```python
from transformers import default_data_collator

train_indices = np.arange(
    len(train_x),
    dtype=np.int64,
)

rng = np.random.default_rng(SEED)
rng.shuffle(train_indices)

dataset = VonpowerChunkDataset(
    train_indices
)

loader = DataLoader(
    dataset,
    batch_size=BATCH_SIZE,
    num_workers=NUM_WORKERS,
    collate_fn=default_data_collator,
    pin_memory=torch.cuda.is_available(),
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
)

scaler = torch.cuda.amp.GradScaler(
    enabled=torch.cuda.is_available()
)

start_time = time.time()

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0
    batch_count = 0

    for batch in loader:
        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }

        optimizer.zero_grad()

        with torch.cuda.amp.autocast(
            enabled=torch.cuda.is_available()
        ):
            outputs = model(**batch)
            loss = outputs.loss

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item()
        batch_count += 1

        if batch_count % 500 == 0:
            print(
                f"Epoch {epoch + 1}, "
                f"Batch {batch_count}, "
                f"Loss: "
                f"{total_loss / batch_count:.4f}"
            )

    print(
        f"Epoch {epoch + 1} finished. "
        f"Average loss: "
        f"{total_loss / batch_count:.4f}"
    )

training_seconds = time.time() - start_time

print(
    f"Training time: "
    f"{training_seconds / 60:.2f} minutes"
)
```

```python
model.eval()

model.save_pretrained(
    DRIVE_OUTPUT
)

tokenizer.save_pretrained(
    DRIVE_OUTPUT
)

print(
    "Final model saved to:"
)
print(DRIVE_OUTPUT)
```
